```python
import pandas as pd
```

```python
# 1. Load the dataset and display the first 10 rows. 
titanic = pd.read_csv(r"C:\Users\sanjn\OneDrive\Desktop\titanic.csv")
titanic.head(10)
```

```python
# 2. Show the shape, column names, and data types. 
print(titanic.info())
print("-"*50)
print(f"Shape : {titanic.shape} \nColumns : {titanic.columns}\n Datatypes : {titanic.dtypes}") 
```

```python
# 3. Find the number of passengers who survived and who did not survive.
survived_data = titanic['survived'].value_counts()
survived_data
```

```python
print("Survived count : ", survived_data.values[1])
print("Not Survived count : ", survived_data.values[0])
```

```python
# 4. Count how many female passengers travelled in 1st class. 
f_1_class = titanic[ (titanic['sex'] == 'female') & (titanic['pclass'] == 1) ]['survived'].count()
print("Count of females travelling in 1st class: ", f_1_class)
```

```python
# 5. Count how many female passengers survived and were under 30 years old. 
f_s_30 = titanic[ (titanic['sex'] == 'female') & (titanic['age'] < 30) & (titanic['survived'] == 1) ]['survived'].count()
print("Count of female passengers survived and were under 30 years old: ", f_s_30)
```

```python
# 6. Count how many male passengers survived and were over 40 years old.
m_s_30 = titanic[ (titanic['sex'] == 'male') & (titanic['age'] > 40) & (titanic['survived'] == 1) ]['survived'].count()
print("Count of male passengers survived and were over 40 years old: ", m_s_30)
```

```python
# 7. Show the count of passengers in each class. 
pclasses = titanic['pclass'].value_counts()
pclasses
```

```python
print("Class First count: ", pclasses.values[1])
print("Class Second count: ", pclasses.values[2])
print("Class Third count: ", pclasses.values[0])
```

```python
# 8. Show the survival count for each class.
pclasses_sur = titanic[(titanic['survived'] == 1)]['pclass'].value_counts()
pclasses_sur
```

```python
# 9. Show the average age of passengers for each class.
print("First class: ", titanic[ titanic['pclass'] == 1]['age'].mean())
print("Second class: ", titanic[ titanic['pclass'] == 2]['age'].mean())
print("Third class: ", titanic[ titanic['pclass'] == 3]['age'].mean())
```

```python
# 10. Replace missing values in the age column with its median. 
titanic.isnull().sum()
```

```python
median = titanic['age'].median()
print("median: ", median)
titanic['age'].fillna(median, inplace=True)
titanic.isnull().sum()
```

```python
# 11. Encode the sex column as numbers (male=1, female=0).
titanic["sex_encode"] = titanic['sex'].map({"male" : 1, "female" : 0})
```

```python
titanic.head(5)
```

```python
# 12. Drop the columns deck and embark_town.
titanic.drop(columns=['deck','embark_town'], inplace=True)
titanic
```

```python
# 13. Find the mean fare paid by survivors vs non-survivors. 
print("Mean fare of survived passengers: ",titanic[titanic['survived'] == 1]['fare'].mean())
print("Mean fare of not survived passengers: ",titanic[titanic['survived'] == 0]['fare'].mean())
```

```python
# 14. Create a new column fare_per_person = fare / number of siblings + spouses + 1. 
titanic['fare_per_person'] = titanic['fare'] / (titanic['sibsp'] + 1)
titanic.head()
```

```python
# 15. Show the top 10 passengers with the highest fare_per_person.
titanic['fare_per_person'].sort_values(ascending=False).head(10)
```
